```python
from langchain_openai import ChatOpenAI
from langchain.messages import HumanMessage, SystemMessage
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm=ChatOpenAI(
    model='deepseek/deepseek-r1-0528:free',
    base_url='https://openrouter.ai/api/v1',
    api_key=os.getenv('OPENROUTER_API')
)
```

```python
messages = [
    SystemMessage(content=
    """
     Role: You are a great mentor with over 20 years of teaching experience in Coding, computerscience and related feilds.
     You only explain or answer for what the user asked , noting more or less.
    when a question related to code is asked to be solved, instead of giving out the solution - you must help them to arrive at the solution.
    # all you do is increase human thinking power - not just provind direct answers or code
    """),
    HumanMessage(content= "Contains Duplicate - help me solve this array related question in java")
]
```

```python
response = llm.invoke(messages)
```

```python
print(response.content)
```

# using groq free models

```python
from openai import OpenAI
import os
```

```python
load_dotenv()
```

```python

client = OpenAI(
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",
)
```

```python

response = client.responses.create(
    model="llama-3.3-70b-versatile",
    input=[
        {
            "role": "system", 
            "content":   """    
                You are a the greatest JAVA teacher and programmer
                when a question related to code is asked to be solved, instead of giving out the solution - you must help them to arrive at the solution.
                 # all you do is increase human thinking power - not just provind direct answers or code.
                 you give a clear explanation when needed making them thinking deep also with small examples.
                 probably give a sample algorithm too - assume they are new to coding
                """},
         {
            "role": "user",
            "content": "Two sum - leetcode problem",
        }
    ],
    max_output_tokens=5000
)
```

```python
op=response.output[1].content[0].text
```

```python
print(op)
```
